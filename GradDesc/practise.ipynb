```python
import pandas as pd
import numpy as np
```

```python
df=pd.read_csv('Social_Network_Ads.csv')
df.head(5)
```

```python
df=pd.get_dummies(df,columns=['Gender'],drop_first=True)
df=df.dropna()
df.head(3)
```

```python
df=df.drop(['User ID'],axis=1)
x=df.drop(['Purchased'],axis=1)
y=df['Purchased']
```

```python
x['Gender_Male']=x['Gender_Male'].astype('int')
x.head(3)
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
```

```python
sc=StandardScaler()
x=sc.fit_transform(x)
```

```python
xtrain,xtest,ytrain,ytest=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
import tensorflow as tf
from tensorflow import keras
from keras import Sequential
from keras.layers import Dense
```

```python
mod=Sequential()
mod.add(Dense(10,activation='relu',input_dim=xtrain.shape[1]))
mod.add(Dense(8,activation='relu'))
mod.add(Dense(1,activation='sigmoid'))
```

```python
mod.summary()
```

```python
#Stocastic GradiDesc
mod.compile(loss='binary_crossentropy',metrics=['accuracy'])
hist=mod.fit(xtrain,ytrain,epochs=5,batch_size=1,validation_split=0.2)
```

```python
#Batch GradiDesc
mod.compile(loss='binary_crossentropy',metrics=['accuracy'])
hist=mod.fit(xtrain,ytrain,epochs=5,batch_size=df.size[0],validation_split=0.2)
```

```python
#Mini-batch GradiDesc
mod.compile(loss='binary_crossentropy',metrics=['accuracy'])
hist=mod.fit(xtrain,ytrain,epochs=5,batch_size=df.size[0]/10,validation_split=0.2)
```
