```python
import numpy as np
import pandas as pd
```

```python
np.random.seed(42) 
cgpa = np.round(np.random.uniform(5.0, 9.5, 100), 1)
profile_score = np.round(np.random.uniform(5.0, 12.0, 100), 1)
lpa = np.round((cgpa * 0.6) + (profile_score * 0.3) + np.random.normal(0, 0.5, 100), 1)
lpa = np.clip(lpa, 2.0, 15.0)  # Keep values within a realistic range

df = pd.DataFrame({
    'cgpa': cgpa,
    'profile_score': profile_score,
    'lpa': lpa
})
df
```

```python
def initialize_parameters():
  
  parameters={
    'w1' : [0.5,0.5,0.5,0.5],
    'b1' : [1,1],
    'w2' : [0.5,0.5],
    'b2' : [1]
  }
  return parameters 
      
```

```python
def forward_multiply(term1,term2,b): 
    return (term1+term2)+b
```

```python
def forward_prop(x1,x2,parameters): 
    o11=forward_multiply(x1*parameters['w1'][0],x2*parameters['w1'][2],parameters['b1'][0])
    o12=forward_multiply(x1*parameters['w1'][1],x2*parameters['w1'][3],parameters['b1'][1])
    o21 = forward_multiply(o11 * parameters['w2'][0], o12 * parameters['w2'][1], parameters['b2'][0])
    return o21, o11, o12
    

```

```python
def update_param(y,x1,x2,parameters,lr) :
    y_hat,o11,o12=forward_prop(x1,x2,parameters)
    dL_dyhat=(y-y_hat)*(-2)
    w2_0 = parameters['w2'][0]
    w2_1 = parameters['w2'][1]
    #layer 1
    parameters['w1'][0] = parameters['w1'][0] - (lr * dL_dyhat * w2_0 * x1)
    parameters['w1'][1] = parameters['w1'][1] - (lr * dL_dyhat * w2_1 * x1)
    parameters['w1'][2] = parameters['w1'][2] - (lr * dL_dyhat * w2_0 * x2)
    parameters['w1'][3] = parameters['w1'][3] - (lr * dL_dyhat * w2_1 * x2)
    
    parameters['b1'][0] = parameters['b1'][0] - (lr * dL_dyhat * w2_0)
    parameters['b1'][1] = parameters['b1'][1] - (lr * dL_dyhat * w2_1)
    
    # Layer 2
    parameters['w2'][0] = parameters['w2'][0] - (lr * dL_dyhat * o11)
    parameters['w2'][1] = parameters['w2'][1] - (lr * dL_dyhat * o12)
    parameters['b2'][0] = parameters['b2'][0] - (lr * dL_dyhat)
    
```

```python
i=2
df.iloc[i,0]
```

```python
def algorithm(df,lr,epoch):
    parameters=initialize_parameters()
    
    for i in range(epoch):
        for j in range(len(df)):
            x1=df.iloc[j,0]
            x2=df.iloc[j,1]
            y=df.iloc[j,2]
            update_param(y,x1,x2,parameters,lr)
    return parameters
```

```python
x=algorithm(df,0.001,10000)
print(x)
```

```python
def predict(x1,x2):
    parameters=x
    o11=parameters['w1'][0]*x1 + parameters['w1'][2]*x2 + parameters['b1'][0]
    o12=parameters['w1'][1]*x1 + parameters['w1'][3]*x2 + parameters['b1'][1]
    y_hat = o11*parameters['w2'][0] + o12*parameters['w2'][1] + parameters['b2'][0]
    return y_hat
```

```python
print(predict(7.7,8.6))
```

```python
from sklearn.metrics import r2_score
```

```python
def r2score() :
    y_true = df['lpa']
    y_pred = [predict(df.iloc[i,0], df.iloc[i,1]) for i in range(len(df))]
    return r2_score(y_true, y_pred)
```
